```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("TelcoCustomerChurn.csv")

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Churn Distribution

plt.figure(figsize=(6,4))
sns.countplot(x='ChurnLabel', data=df)

plt.title('Customer Churn Distribution')
plt.show()


# Contract Type vs Churn

plt.figure(figsize=(8,5))
sns.countplot(x='Contract', hue='ChurnLabel', data=df)

plt.title('Contract Type vs Churn')
plt.xticks(rotation=15)
plt.show()


# Monthly Charges vs Churn

plt.figure(figsize=(8,5))
sns.boxplot(x='ChurnLabel', y='MonthlyCharge', data=df)

plt.title('Monthly Charges vs Churn')
plt.show()


# Tenure vs Churn

plt.figure(figsize=(8,5))
sns.boxplot(x='ChurnLabel', y='TenureinMonths', data=df)

plt.title('Tenure vs Churn')
plt.show()


# Customer Status

plt.figure(figsize=(8,5))
sns.countplot(x='CustomerStatus', data=df)

plt.title('Customer Status Distribution')
plt.xticks(rotation=20)
plt.show()
```

Key Insights

1. Approximately 26.5% of customers have churned.

2. Customers with Month-to-Month contracts show the highest churn rate.

3. Customers with longer tenure are less likely to churn.

4. Customers who churn tend to have higher monthly charges.

5. Long-term contracts help improve customer retention.
